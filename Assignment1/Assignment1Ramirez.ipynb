# Part I. Describe: How bad is it, really?

#### A) For any provided promised delivery time and each pair of specified zone and time block, write a function that returns the percentage of late orders.

### Logic for Part 1
1. **Pseudocode** (**A:**): 
    - When we enter our variables, they need to be passed into the delivery_times function, which returns a array of values, we need to determine which values are over our threshold (45 minutes), in the case of the hardcoded promise. 
    - We can iterate over the array, using if/else logic to increment a (**lateness**) variable (a list that has +1 added for every value greater than promise)
    - We can ge the total count using len of the times varible.
    - We then return lateness over len times as the outout of the function to return the percentage of late deliveries per zone.
2. **Pseudocode** (Handling All inputs):
    - When we input the zone as all, of which there are three, we need delivery time to loop through zone one at a time going through value, collecting the total in addtion to all the late delivries while still using a fixed value for promist and timeblock.
    - *revised* Apply the same logic as above to timezone, when building out the function, using conditional logic start with the assumption that it's all by default, else use specififed value.
3. **Pseudocode** (**B:** Ranking the pairs):
    - To determine the rank of pairs in descending order of late arrival rate, we need to capture the output of each permutation (12) and then order them from greatest to least.
    - We could run each combination of LateOrders in a for loo, capturing the output (a combination of the zone,timeblock and promise) in a list and then sorting it. 

4. **Pseudocode** (**C:** returning the average delivery time):
    - We can reuse blocks of the code from LateOrders.
    - We can repurpose the logic that enables us to use (all) and a specific zone, along with the loop that iterates through each permutation. 
    - Moreover, from the sample code, we can use the time.mean method to capture the average when the input is a non-all value.
    - In order to handle the all:one, all:all, and one:all scenarios we need caputure a running total of the minutes across each combination and then divide by the total inputs (orders) 

````python
#Code for part A

def LateOrders(zone, time_block, promise=PROMISE, show_statement=True):
    selected_zones = ZONES if zone == "all" else [zone]
    selected_time_blocks = TIME_BLOCKS if time_block == "all" else [time_block]

    late_count = 0
    total_count = 0

    for selected_zone in selected_zones:
        for selected_time_block in selected_time_blocks:
            times = delivery_times(selected_zone, selected_time_block, promise)
            total_count += len(times)

            for duration in times:
                if duration > promise:
                    late_count += 1

    percentage = 100 * late_count / total_count
    if show_statement:
        zone_label = "all zones" if zone == "all" else zone
        time_block_label = "all time blocks" if time_block == "all" else time_block
        print(
            f"For zone {zone_label} and time block {time_block_label}, "
            f"with a promise of {promise} minutes, {percentage:.2f}% of orders "
            "will be late."
        )
    return percentage


# Specific zone and time block
LateOrders("Far West", "Fri/Sat eve")

# All zones, one time block
LateOrders("all", "Fri/Sat eve")

# One zone, all time blocks
LateOrders("Far West", "all")

# All zones and all time blocks
LateOrders("all", "all")
````

````python
#Code for Part B
ranked_pairs = []

for zone in ZONES:
    for time_block in TIME_BLOCKS:
        late_rate = LateOrders(
            zone,
            time_block,
            promise=45,
            show_statement=False,
        )
        ranked_pairs.append((late_rate, zone, time_block))

ranked_pairs.sort(reverse=True)

for rank, (late_rate, zone, time_block) in enumerate(ranked_pairs, start=1):
    print(
        f"{rank}. {zone} / {time_block}: "
        f"{late_rate:.2f}% late"
    )
````

````python
#Code for part C
def AverageDeliveryTime (zone, time_block, promise=PROMISE):
    selected_zones = ZONES if zone == "all" else [zone]
    selected_time_blocks = TIME_BLOCKS if time_block == "all" else [time_block]

    TotalMinutes = 0
    TotalOrders = 0

    for selected_zone in selected_zones:
        for selected_time_block in selected_time_blocks:
            times = delivery_times(selected_zone, selected_time_block, promise)
            TotalMinutes += times.sum()
            TotalOrders +=len(times)
    return TotalMinutes / TotalOrders
````

````python
print("Specific zone and specific time block:")
print(f"{AverageDeliveryTime('Far West', 'Fri/Sat eve', promise=promise):.1f} minutes")

print("All zones and a specific time block:")
print(f"{AverageDeliveryTime('all', 'Fri/Sat eve', promise=promise):.1f} minutes")

print("A specific zone and all time blocks:")
print(f"{AverageDeliveryTime('Far West', 'all', promise=promise):.1f} minutes")

print("All zones and all time blocks:")
print(f"{AverageDeliveryTime('all', 'all', promise=promise):.1f} minutes")
````
