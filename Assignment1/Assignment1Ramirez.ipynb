````python
#Code for part A

def LateOrders(zone, time_block, promise=PROMISE, show_statement=True):
    selected_zones = ZONES if zone == "all" else [zone]
    selected_time_blocks = TIME_BLOCKS if time_block == "all" else [time_block]

    late_count = 0
    total_count = 0

    for selected_zone in selected_zones:
        for selected_time_block in selected_time_blocks:
            times = delivery_times(selected_zone, selected_time_block, promise)
            total_count += len(times)

            for duration in times:
                if duration > promise:
                    late_count += 1

    percentage = 100 * late_count / total_count
    if show_statement:
        zone_label = "all zones" if zone == "all" else zone
        time_block_label = "all time blocks" if time_block == "all" else time_block
        print(
            f"For zone {zone_label} and time block {time_block_label}, "
            f"with a promise of {promise} minutes, {percentage:.2f}% of orders "
            "will be late."
        )
    return percentage


# Specific zone and time block
LateOrders("Far West", "Fri/Sat eve")

# All zones, one time block
LateOrders("all", "Fri/Sat eve")

# One zone, all time blocks
LateOrders("Far West", "all")

# All zones and all time blocks
LateOrders("all", "all")
````

````python
#Code for Part B
ranked_pairs = []

for zone in ZONES:
    for time_block in TIME_BLOCKS:
        late_rate = LateOrders(
            zone,
            time_block,
            promise=45,
            show_statement=False,
        )
        ranked_pairs.append((late_rate, zone, time_block))

ranked_pairs.sort(reverse=True)

for rank, (late_rate, zone, time_block) in enumerate(ranked_pairs, start=1):
    print(
        f"{rank}. {zone} / {time_block}: "
        f"{late_rate:.2f}% late"
    )
````

````python
#Code for part C
def AverageDeliveryTime (zone, time_block, promise=PROMISE):
    selected_zones = ZONES if zone == "all" else [zone]
    selected_time_blocks = TIME_BLOCKS if time_block == "all" else [time_block]

    TotalMinutes = 0
    TotalOrders = 0

    for selected_zone in selected_zones:
        for selected_time_block in selected_time_blocks:
            times = delivery_times(selected_zone, selected_time_block, promise)
            TotalMinutes += times.sum()
            TotalOrders +=len(times)
    return TotalMinutes / TotalOrders

````

````python
print("Specific zone and specific time block:")
print(f"{AverageDeliveryTime('Far West', 'Fri/Sat eve', promise=PROMISE):.1f} minutes")

print("All zones and a specific time block:")
print(f"{AverageDeliveryTime('all', 'Fri/Sat eve', promise=PROMISE):.1f} minutes")

print("A specific zone and all time blocks:")
print(f"{AverageDeliveryTime('Far West', 'all', promise=PROMISE):.1f} minutes")

print("All zones and all time blocks:")
print(f"{AverageDeliveryTime('all', 'all', promise=PROMISE):.1f} minutes")
````

# Part 2. Exploring the best promise

### Logic for Part 2
1. **Pseudocode** (**A:**): 
    - Refunds cost $10, 1.8 future orders are lost as a result of a delayed order, and the margin on each order is $9.
    - 10 + (1.8*9) = 26.20, total cost of one late order
    - LateCost = NumeberOfLateOrders * (Refund + Churn * Margin)
2. **Pseudocode** (**B:**):

   **Pseudocode** (**Choosing the range of promises**):
    - The function only evaluates the promises it is given, so the range has to be settled before the final run. It needs to be wide enough that each pair's best promise sits inside the list rather than on its edge, since a winner on an endpoint only tells us it was the best of what we tried.
    - We start with a wide list, 20 to 90 in steps of 5. Below 20 minutes nearly every order would be late, and by 90 demand is close to zero, so the peak for any pair should fall in between.
    - We loop over every zone and time block (12 pairs), call the function on each with the same wide list, and record each pair's best promise. All 12 are tested because each pair has its own peak, and the pair with the highest late rate at 45 minutes is not guaranteed to have the highest optimal promise.
    - If any pair's best promise is 20 or 90, the list is widened and the sweep is rerun.
    - The sweep returned best promises between 45 and 55 minutes, with no pair on an edge.
    - The final list pads the lowest and highest winners by 10 minutes and steps by 1 minute, giving 35 to 65 (list(range(35, 66))). The 1-minute step pins down the exact minute, and 45 sits inside the list, so each pair's profit at today's promise is available as a baseline.
    - The final run repeats the loop over all 12 pairs with this one shared list, and the same edge check applies: no winner should be 35 or 65.
    - As a robustness check, the final list was rerun with seeds 1 to 5. Every winner across all pairs and seeds fell between 43 and 58, inside the list with room on both sides.

   **Pseudocode** (**Implementation of the function**):
    - Inputs are the zone, the time block, the list of promises, the costs (COSTS by default) and a seed. The outputs are the promise with the highest net profit and that net profit. An optional return_results` flag also returns the (promise, net profit) pair for every candidate, which is useful for checking the shape of the profit curve.
    - The function handles one zone and time block per call. Covering all 12 pairs means calling it in a loop over ZONES and TIME_BLOCKS, so the loop over pairs sits outside the function and the loop over promises sits inside it.
    - For each promise in the list, we call delivery_times with that promise. The promise affects demand, so every candidate needs a fresh call and the results from one promise cannot be reused for another. The same seed is passed each time so the comparison is consistent and the notebook reproduces.
    - From each call, the number of orders is the length of the array, and the number of late orders is the count of delivery times greater than that same promise.
    - Net profit is the margin on every order minus the late cost, orders * margin - LateCost(late). Late orders are still delivered and paid for, and the late cost already covers the refund and the churned future margin, so removing the margin on late orders as well would count that loss twice. LateCost from Part 2(a) is reused, so the cost per late order is defined in one place.
    - We keep a running best. If a candidate's net profit is higher than the best so far (or nothing has been recorded yet), it replaces the stored promise and profit. Ties keep the shorter promise.
    - The function never edits the list it receives. Any refinement, such as a finer step around a winner, is done by calling it again with a new list, and the same parameter becomes the range input in the Part III app.

   **Assumptions and limits**:
    - Net profit counts margin on all orders, as described above.
    - Dollar figures are totals over the four simulated weeks.
    - The results come from a simulator, so winners shift by a few minutes from seed to seed, and the recommended promises should be read as approximate. The pairs with the largest gain over 45 minutes, particularly Far West / Fri/Sat eve, are the ones where the recommendation is clearest.


````python
def LateCost(numlate, costs=COSTS):
    cost_per_late_order = (
        costs["refund"]
        + costs["churn_orders"] * costs["margin"]
    )
    return numlate * cost_per_late_order

LateCost(10)
````

````python
def BestPromise(zone, time_block, promises, costs=COSTS, seed=1, return_results=False):
    results = []
    best_promise = None
    best_profit = None

    for promise in promises:
        times = delivery_times(zone, time_block, promise, seed=seed)

        orders = len(times)
        late = int(np.sum(times > promise))

        # margin on every order, minus the full cost of each late one
        profit = orders * costs["margin"] - LateCost(late, costs)
        results.append((promise, profit))

        if best_profit is None or profit > best_profit:
            best_profit = profit
            best_promise = promise

    if return_results:
        return best_promise, best_profit, results
    return best_promise, best_profit
````

````python
# wide_list = list(range(20, 95, 5))
FinalList = list(range(35, 70))
BestPromise("Far West", "Fri/Sat eve", FinalList)
````

````python
wide_list = list(range(20, 95, 5))
FinalList = list(range(35, 66))

for zone in ZONES:
    for time_block in TIME_BLOCKS:
        wide_best, _ = BestPromise(zone, time_block, wide_list)
        best, profit = BestPromise(zone, time_block, FinalList)
        _, base = BestPromise(zone, time_block, [45])
        print(f"{zone} / {time_block}: wide={wide_best}, best={best} min, "
              f"${profit:,.2f} (vs ${base:,.2f} at 45, +${profit - base:,.2f})")
````

## Part 2(b): Review Guide

### Decision
For one zone and one time block, choose the candidate promised delivery time that produces the highest net profit. The function evaluates a supplied list of promises; it does not need to create that list itself.

### Choose Candidate Promises
- Choose a practical range based on delivery-time results and business constraints.
- A 5-minute step is a reasonable first pass. Larger steps are quicker but may skip a better candidate; smaller steps give more precision.
- If the best candidate is at either endpoint, expand the range and check again. If it is in the middle, try finer steps around it.
- The result is the best promise among those tested, not necessarily the best possible promise.

### Evaluate Each Candidate
For each promise in the list:
1. Get delivery results for the specified zone and time block.
2. Use `COSTS` and the assignment's rules to calculate net profit: customer profit collected minus late costs.
3. Keep the result with its promise, such as `(net_profit, promise)`.

Because the function handles one zone/time-block pair, its result does not need to repeat those inputs. If comparing results across pairs later, include the zone and time block too. Keeping all candidate results is useful for explaining or charting the choice; sorting them is optional if the function only needs to return the winner.

### Assumptions to Check
- Confirm exactly how `COSTS` defines customer profit and late costs.
- If customer profit is fixed and longer promises only reduce late costs, the longest candidate may always win. Check whether the cost model creates a counterbalancing trade-off.
- Delivery durations are simulated. If a fresh random sample is generated for each candidate, differences in profit may partly reflect simulation noise. Check whether the comparison can use common delivery data or repeated simulations.
- Decide how to handle tied net-profit results.
